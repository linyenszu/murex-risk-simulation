## Black-Scholes (or Black-Scholes-Merton) Model
The Black-Scholes (or Black-Scholes-Merton) formula is a mathematical model used to estimate the theoretical fair price of European-style options. It assumes that the underlying asset price follows the Geometric Brownian Motion (GBM) that you have been modeling. For a non-dividend-paying stock, the model calculates the option price based on current stock price, strike price, time to expiration, risk-free interest rate, and volatility. The Black-Scholes EquationsFor a Call Option ($C$) and a Put Option ($P$), the formulas are:$$C(S_t, t) = S_t N(d_1) - K e^{-r(T-t)} N(d_2)$$$$P(S_t, t) = K e^{-r(T-t)} N(-d_2) - S_t N(-d_1)$$Where $d_1$ and $d_2$ are defined as:$$d_1 = \frac{\ln(S_t / K) + \left(r + \frac{1}{2}\sigma^2\right)(T-t)}{\sigma \sqrt{T-t}}$$$$d_2 = d_1 - \sigma \sqrt{T-t} = \frac{\ln(S_t / K) + \left(r - \frac{1}{2}\sigma^2\right)(T-t)}{\sigma \sqrt{T-t}}$$Variable Definitions:
* $S_t$ = Current spot price of the underlying asset
* $K$ = Strike price (exercise price) of the option
* $T - t$ = Time remaining until expiration (expressed in years)
* $r$ = Risk-free continuous interest rate
* $\sigma$ = Annualized volatility of the asset's returns
* $N(x)$ = The cumulative distribution function (CDF) of a standard normal distribution $N(0,1)$. It represents the probability that a random variable is less than or equal to $x$.
#### Adding Dividend Yield ($q$)
If you are expanding this to fit your updated GBM model that includes a continuous dividend yield ($q$), the formula adjusts by discounting the current stock price term ($S_t$ becomes $S_t e^{-q(T-t)}$):$$C = S_t e^{-q(T-t)} N(d_1) - K e^{-r(T-t)} N(d_2)$$$$P = K e^{-r(T-t)} N(-d_2) - S_t e^{-q(T-t)} N(-d_1)$$With the modified $d_1$ term incorporating $q$:$$d_1 = \frac{\ln(S_t / K) + \left(r - q + \frac{1}{2}\sigma^2\right)(T-t)}{\sigma \sqrt{T-t}}$$Python Code Implementation

## 6. Sanity checks

These checks make the notebook suitable for a production project review: no missing risk values, stock Delta equals quantity, and option Vega is finite.


```python
assert instrument_risk[GREEK_COLUMNS].notna().all().all(), "Missing Greek values"

stock_rows = instrument_risk["InstrumentType"].eq("Stock")
assert np.allclose(
    instrument_risk.loc[stock_rows, "Delta"],
    instrument_risk.loc[stock_rows, "Quantity"],
), "Stock Delta should equal quantity"

option_rows = instrument_risk["InstrumentType"].eq("European Option")
assert np.isfinite(instrument_risk.loc[option_rows, "Vega"]).all(), "Option Vega must be finite"

print("Risk checks passed")
```

## 8. Risk visualization

Plot the largest absolute Delta contributors to quickly identify directional exposure concentration.


```python
import matplotlib.pyplot as plt

plot_df = instrument_risk.assign(AbsDelta=instrument_risk["Delta"].abs()).sort_values("AbsDelta", ascending=False)
ax = plot_df.plot.bar(x="PositionID", y="AbsDelta", figsize=(11, 5), legend=False, title="Largest Delta Contributors")
ax.set_xlabel("Position")
ax.set_ylabel("Absolute Delta")
plt.tight_layout()
```
